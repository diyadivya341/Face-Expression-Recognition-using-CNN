```python
import tensorflow as tf
import keras
import warnings
warnings.simplefilter('ignore')
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

```python
train_datagen = ImageDataGenerator(rescale = 1/255,
                                   shear_range = 0.2,
                                   zoom_range = 0.2,
                                  validation_split=0.2)
```

```python
training_set = train_datagen.flow_from_directory(
    r"C:\Users\divya\Downloads\archive\Dataset(Final)\train",  # Set the correct path to your dataset
    target_size=(64, 64),   # Resize images to fit the model input
    batch_size=32,
    class_mode='categorical')
```

```python
training_set.class_indices
```

```python
test_datagen = ImageDataGenerator(rescale = 1/255)

test_set = test_datagen.flow_from_directory(r"C:\Users\divya\Downloads\archive\Dataset(Final)\test",
                                           target_size = (64,64),
                                            class_mode = "categorical")
```

```python
test_set.class_indices
```

**Modelling - Convolution Neural Network**

**Initialising the CNN**

```python
from keras.models import Sequential
Classifier = Sequential()
```

**Step 1 Convolution**

```python
from keras.layers import Conv2D
Classifier.add(Conv2D(input_shape=[64, 64, 3],
                      filters=32, kernel_size=3,activation='relu'))
```

```python
Classifier.add(Conv2D(input_shape=[64, 64, 3],
                      filters=32, kernel_size=3,activation='relu'))
```

**Step 2 Max-Polling**

```python
from keras.layers import MaxPooling2D
Classifier.add(MaxPooling2D(pool_size=2,strides=(2,2)))
```

**Step 3: Flattening**

```python
from keras.layers import Flatten
Classifier.add(Flatten())
```

**Step 4: Full Connection**

```python
from keras.layers import Dense

# hidden layer with 128 neurons 
Classifier.add(Dense(units=128, activation='relu'))

# Output layer with 5 neurons
Classifier.add(Dense(units = 5, activation = 'softmax'))
```

**Training the CNN Model with train data & testing the model with test data**

```python
Classifier.compile(optimizer = 'adam',
                   loss = 'categorical_crossentropy',
                   metrics = ['accuracy'])
```

```python
Classifier.fit(x=training_set, validation_data=test_set, epochs=25)
```

**Evalution**
 - **Making a single prediction**

```python
import numpy as np
from PIL import Image
```

```python
import numpy as np
from PIL import Image

# Load the image
img_path = (r"C:\Users\divya\Downloads\archive\Dataset(Final)\val\Surprised_human_face\flipped_cauchandsome-man-blue-surprised-pointing-finger-side_1368-95322.jpg_face_0.jpg")
test_image = Image.open(img_path)

# Preprocess
test_image = test_image.resize((64, 64))
test_image = np.array(test_image)
test_image = test_image / 255.0  # Normalize
test_image = np.expand_dims(test_image, axis=0)  # Add batch dimension

# Predict
result = Classifier.predict(test_image)

# Get predicted class index
predicted_class_index = np.argmax(result)

# Get class label names
class_labels = list(training_set.class_indices.keys())
predicted_label = class_labels[predicted_class_index]

# Show prediction
print("Predicted Class:", predicted_label)
```
